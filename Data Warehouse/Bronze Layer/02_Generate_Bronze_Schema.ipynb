```python
import pandas as pd

base_path=r"C:\Users\Ayush\Git Repo\Stack Over Flow Survey Data Engineering Project\Stack Overflow survey\\"
years=[2021,2022,2023,2024,2025]

for year in years:
    real_path=base_path+f"survey_results_public_{year}.csv"
    df=pd.read_csv(real_path,nrows=0)
    
    sql_columns=[f"[{col}] NVARCHAR (MAX)" for col in df.columns]
    # sql_columns.append('[SurveyYear] int')
    
    sql_columns_final=",\n".join(sql_columns)
    query=f"CREATE TABLE Bronze.Survey_{year} (\n{sql_columns_final}\n);"
    print(f"-- Table for {year} --")
    print(query)
    print('go')
    print('\n\n')
```
